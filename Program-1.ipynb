```python
import numpy as np
import pandas as pd

# NumPy operations
arr = np.array([78, 92, 65, 84, 71])

print("array:", arr)
print("mean:", arr.mean())
print("reshaped 2D:\n", arr.reshape(5, 1))
print("sliced(index 1 to 3):", arr[1:4])

# Pandas operations
data = {
    "Student": ['Arjun', 'Meera', 'Vikram', 'Neha'],
    "Attendance": [85, 92, 78, 95],
    "Score": [76, 89, 68, 94],
}

df = pd.DataFrame(data)

print("\nFirst two rows:\n", df.iloc[0:2])

print("\nScore column:\n", df['Score'])

print("\nRows where Score > 80:\n", df[df['Score'] > 80])
```
