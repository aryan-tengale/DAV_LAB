```python
import pandas as pd

data = {
    'Employee': ['Rahul', 'Sneha', 'Arjun', 'Meera', 'Kiran', 'Divya'],
    'Department': ['IT', 'HR', 'Finance', 'Marketing', 'IT', 'Finance'],
    'Salary': [65000, 52000, 70000, 58000, 62000, 75000],
    'Performance': [85, 72, 91, 68, 78, 88],
    'Level': ['Excellent', 'Good', 'Excellent', 'Average', 'Good', 'Excellent']
}

df = pd.DataFrame(data)

df.to_csv('employees.csv', index=False)

print("employees.csv created successfully!")
```

```python
import pandas as pd

# Load the dataset directly from your working directory
df = pd.read_csv('employees.csv')

# Display summary statistics
print("Summary Statistics:\n", df.describe())

# Check for missing values and duplicate rows
print('\nMissing values:\n', df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())

# Salary DataFrame based on departments in employees.csv
Salary = pd.DataFrame({
    'Department': ['IT', 'HR', 'Finance', 'Marketing'],
    'Bonus': [50000, 35000, 45000, 40000]
})

# Merge Bonus details with employee dataset
merged = pd.merge(df, Salary, on='Department', how='left')
print("\nMerged Data Sample:\n", merged.head())

# Calculate average performance score grouped by Department
print('\nAverage Performance by Department:\n',
      df.groupby('Department')['Performance'].mean())

# Create and display a pivot table
# Counting employees by Department and Performance Level
pivot = pd.pivot_table(
    df,
    values='Performance',
    index='Department',
    columns='Level',
    aggfunc='count'
)

print('\nEmployee Count by Department and Performance Level:\n', pivot)
```
